# Comparación: Train/Test Split vs Stratified K-Fold (Titanic)
Este notebook carga el dataset de Titanic, realiza preprocessing básico y compara ambos enfoques.

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split, StratifiedKFold, cross_val_score
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import OneHotEncoder
```

```python
# Cargar dataset
df = pd.read_csv('\data\phpMYEkMl.csv')

df.head()
```

```python
df.info()
```

```python
# Variables
target = 'survived'
X = df.drop(columns=['survived', 'name', 'ticket', 'cabin'])
y = df[target]

# Seleccionar columnas numéricas y categóricas
cat_cols = ['sex', 'embarked']
num_cols = ['age', 'fare', 'pclass', 'sibsp', 'parch']

# Pipelines
num_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

cat_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(handle_unknown='ignore'))
])

preprocessor = ColumnTransformer([
    ('num', num_pipeline, num_cols),
    ('cat', cat_pipeline, cat_cols)
])

model = Pipeline([
    ('preprocessor', preprocessor),
    ('classifier', LogisticRegression(max_iter=1000))
])
```

```python
# Train/Test Split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)

model.fit(X_train, y_train)
y_pred = model.predict(X_test)

acc_split = accuracy_score(y_test, y_pred)
print("Accuracy (Train/Test Split):", acc_split)
```

```python
# Stratified K-Fold
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

scores = cross_val_score(model, X, y, cv=skf, scoring='accuracy')

print("Accuracy por fold:", scores)
print("Accuracy promedio:", scores.mean())
```

## Conclusión
- Train/Test Split es más simple pero depende de una sola partición.
- Stratified K-Fold es más robusto al promediar múltiples particiones manteniendo proporciones de clases.
